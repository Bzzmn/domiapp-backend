```python
from unstructured.documents.elements import (
    CompositeElement, 
    Table, 
    FigureCaption, 
    Title, 
    NarrativeText,
    ElementMetadata
)
from typing import Optional, Type, List, Tuple

def find_previous_element(elements: List, current_index: int) -> Optional[Tuple[int, any]]:
    """
    Busca el elemento anterior más relevante dentro de los 3 elementos previos
    siguiendo el orden de prioridad: FigureCaption > Title > NarrativeText
    """
    # Definir orden de prioridad
    priority_types = [FigureCaption, Title, NarrativeText]
    
    # Buscar en los últimos 3 elementos
    start_index = max(0, current_index - 3)
    previous_elements = elements[start_index:current_index]
    
    # Buscar según prioridad
    for element_type in priority_types:
        for i, element in enumerate(reversed(previous_elements)):
            if isinstance(element, element_type):
                return (current_index - i - 1, element)
    
    return None

def extract_table_with_priority_previous(
    composite_element: CompositeElement
) -> List[CompositeElement]:
    """
    Extrae cada tabla junto con el elemento anterior más relevante 
    dentro de los 3 elementos previos
    """
    result = []
    elements = composite_element.metadata.orig_elements
    
    for i, element in enumerate(elements):
        if isinstance(element, Table):
            # Buscar el elemento anterior según prioridad
            previous = find_previous_element(elements, i)
            
            if previous:
                prev_index, previous_element = previous
                
                # Crear metadata
                metadata = ElementMetadata()
                metadata.page_number = getattr(element.metadata, "page_number", None)
                metadata.orig_elements = [previous_element, element]
                metadata.filetype = getattr(composite_element.metadata, "filetype", None)
                metadata.languages = getattr(composite_element.metadata, "languages", None)
                metadata.previous_element_type = type(previous_element).__name__
                metadata.previous_element_distance = i - prev_index
                metadata.table_position = i
                
                # Crear nuevo CompositeElement
                new_composite = CompositeElement(
                    text=f"{previous_element.text}\n\n{element.text}",
                    metadata=metadata
                )
                result.append(new_composite)
    
    return result
```

```python
# separate tables from texts
tables = []
texts = []

for chunk in chunks:
    texts.append(chunk)
    if isinstance(chunk, CompositeElement):
        pairs = extract_table_with_priority_previous(chunk)
        tables.extend(pairs)

print(len(texts))
print(len(tables))
```

```python
from unstructured.documents.elements import (
    CompositeElement, 
    FigureCaption, 
    Title, 
    NarrativeText
)
from typing import Optional, Union, List

def extract_title_from_composite(
    composite: CompositeElement
) -> Optional[str]:
    """
    Extrae el texto del título de un CompositeElement.
    
    El título se define como el primer elemento del CompositeElement
    que sea de tipo FigureCaption, Title o NarrativeText.
    
    Args:
        composite: El CompositeElement del que extraer el título
        
    Returns:
        El texto del título o None si no se encuentra
    """
    # Verificar que sea un CompositeElement
    if not isinstance(composite, CompositeElement):
        return None
    
    # Obtener los elementos originales
    elements = composite.metadata.orig_elements
    
    # Verificar que haya elementos
    if not elements or len(elements) == 0:
        return None
    
    # Obtener el primer elemento
    first_element = elements[0]
    
    # Verificar que sea uno de los tipos de título
    if isinstance(first_element, (FigureCaption, Title, NarrativeText)):
        return first_element.text
    
    return None

def extract_titles_from_composites(
    composites: List[CompositeElement]
) -> List[str]:
    """
    Extrae los títulos de una lista de CompositeElements.
    
    Args:
        composites: Lista de CompositeElements
        
    Returns:
        Lista de textos de títulos (sin None)
    """
    titles = []
    
    for composite in composites:
        title = extract_title_from_composite(composite)
        if title:
            titles.append(title)
    
    return titles

# Ejemplo de uso
"""
# Para un solo CompositeElement
title = extract_title_from_composite(composite_element)
if title:
    print(f"Título encontrado: {title}")
else:
    print("No se encontró título en este CompositeElement")

# Para una lista de CompositeElements
chunks = [...]  # Lista de chunks o elementos
composite_chunks = [chunk for chunk in chunks if isinstance(chunk, CompositeElement)]
titles = extract_titles_from_composites(composite_chunks)
print(f"Se encontraron {len(titles)} títulos:")
for i, title in enumerate(titles, 1):
    print(f"{i}. {title}")
"""

# Función adicional para obtener el título con información de tipo
def extract_title_with_type(
    composite: CompositeElement
) -> Optional[dict]:
    """
    Extrae el título con información adicional sobre su tipo.
    
    Args:
        composite: El CompositeElement del que extraer el título
        
    Returns:
        Diccionario con texto del título y su tipo, o None si no se encuentra
    """
    # Verificar que sea un CompositeElement
    if not isinstance(composite, CompositeElement):
        return None
    
    # Obtener los elementos originales
    elements = composite.metadata.orig_elements
    
    # Verificar que haya elementos
    if not elements or len(elements) == 0:
        return None
    
    # Obtener el primer elemento
    first_element = elements[0]
    
    # Verificar que sea uno de los tipos de título
    if isinstance(first_element, (FigureCaption, Title, NarrativeText)):
        element_type = first_element.__class__.__name__
        return {
            "text": first_element.text,
            "type": element_type,
            "page_number": getattr(first_element.metadata, "page_number", None)
        }
    
    return None
```

```python
import re
from unstructured.documents.elements import (
    CompositeElement, 
    Text,
    NarrativeText,
    Title
)
from typing import Union, List, Optional

def extract_structural_texts(
    elements: Union[CompositeElement, Text, List[Union[CompositeElement, Text]]]
) -> str:
    """
    Extrae textos que coinciden con patrones estructurales (capítulos, párrafos, artículos y cuadros)
    y devuelve un texto concatenado con todos los resultados.
    
    Args:
        elements: Un CompositeElement individual o una lista de elementos
        
    Returns:
        String con todos los textos concatenados con saltos de línea
    """
    # Convertir un solo elemento a lista si es necesario
    if not isinstance(elements, list):
        elements = [elements]
    
    # Definir patrones regex - Ahora sin IGNORECASE para capítulos, párrafos y artículos
    patterns = [
        re.compile(r'\b(CAPITULO|CAPÍTULO)\s+(?:[IVX]+|\d+(?:\.\d+)?)'),
        re.compile(r'\b(PARRAFO|PÁRRAFO)\s+\d+(?:\.\d+)?'),
        re.compile(r'\b(ART(?:Í|I)CULO)\s+\d+(?:\.\d+)?(?:\.\d+)?'),
        re.compile(r'\b(?:CUADRO|TABLA|Cuadro|Tabla)\s+(?:N°|N|Nº|No\.|No|Núm\.|Núm)\s*\d+(?:\.[A-Z])?', re.IGNORECASE)
    ]
    
    # Lista para almacenar resultados
    results = []
    
    # Procesar cada elemento
    for element in elements:
        # Obtener texto del elemento
        if isinstance(element, CompositeElement):
            text = element.text
        else:
            text = element.text if hasattr(element, "text") else str(element)
        
        # Buscar coincidencias para cada patrón en el texto
        for pattern in patterns:
            matches = pattern.finditer(text)
            
            for match in matches:
                # Obtener texto coincidente y añadirlo a resultados
                match_text = match.group(0)
                results.append(match_text)
    
    # Concatenar resultados con saltos de línea
    return "\n".join(results) if results else ""
```

```python
from typing import List, Optional

def zip_text_lists(list1: List[str], list2: List[str], separator: str = "\n\n") -> List[str]:
    """
    Combina dos listas de textos por índice.
    
    Args:
        list1: Primera lista de textos
        list2: Segunda lista de textos
        separator: Separador a usar entre los textos combinados
        
    Returns:
        Lista de textos combinados
    """
    # Determinar la longitud mínima entre las dos listas
    min_length = min(len(list1), len(list2))
    
    # Combinar los textos por índice
    result = []
    for i in range(min_length):
        # Verificar que ambos textos existan
        if list1[i] and list2[i]:
            combined_text = f"{list1[i]}{separator}{list2[i]}"
        elif list1[i]:
            combined_text = list1[i]
        elif list2[i]:
            combined_text = list2[i]
        else:
            combined_text = ""
        
        result.append(combined_text)
    
    # Manejar los elementos adicionales de la lista más larga
    if len(list1) > min_length:
        result.extend(list1[min_length:])
    elif len(list2) > min_length:
        result.extend(list2[min_length:])
    
    return result
```

```python
import os
import uuid
from typing import List, Dict, Any, Optional
import pickle

# Importaciones para Pinecone
from pinecone import Pinecone
from langchain_pinecone import PineconeVectorStore
from langchain_openai import OpenAIEmbeddings

# Importaciones para Redis
from redis import Redis
from langchain_community.storage import RedisStore

# Importaciones de LangChain
from langchain_core.documents import Document
from langchain.retrievers.multi_vector import MultiVectorRetriever

# Configuración de variables de entorno (asumiendo que ya están en .env)
PINECONE_API_KEY = os.environ.get("PINECONE_API_KEY")
PINECONE_ENV = os.environ.get("PINECONE_ENVIRONMENT", "gcp-starter")
REDIS_URL = os.environ.get("REDIS_URL")
REDIS_NAMESPACE = os.environ.get("REDIS_NAMESPACE", "docs")

# Inicializar Pinecone
vectorstore = PineconeVectorStore(
    pinecone_api_key=PINECONE_API_KEY,
    index_name=PINECONE_INDEX,
    embedding=OpenAIEmbeddings(),
)

doc_store = RedisStore(
    redis_url=REDIS_URL,
    namespace=REDIS_NAMESPACE,
)

class PickleMultiVectorRetriever(MultiVectorRetriever):
    """MultiVectorRetriever con soporte para deserialización automática."""
    
    def invoke(self, query, config=None, **kwargs):
        # Obtener los IDs de documentos del vectorstore
        ids = self.vectorstore.similarity_search(query, **kwargs.get("search_kwargs", {}))
        # Obtener los documentos del docstore por IDs
        docs = self.docstore.mget([d.metadata[self.id_key] for d in ids])
        # Deserializar los documentos
        return [pickle.loads(doc) if doc else None for doc in docs]


retriever = PickleMultiVectorRetriever(
    vectorstore=vectorstore,
    docstore=doc_store,
    id_key=id_key,
)
```

```python
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_openai import ChatOpenAI
from base64 import b64decode


def parse_docs(docs):
    """Split base64-encoded images and texts"""
    b64 = []
    text = []
    for doc in docs:
        try:
            b64decode(doc)
            b64.append(doc)
        except Exception as e:
            text.append(doc)
    return {"images": b64, "texts": text}


def build_prompt(kwargs):

    docs_by_type = kwargs["context"]
    user_question = kwargs["question"]

    context_text = ""
    if len(docs_by_type["texts"]) > 0:
        for text_element in docs_by_type["texts"]:
            context_text += text_element.text

    # construct prompt with context (including images)
    prompt_template = f"""
    Answer the question based only on the following context, which can include text, tables, and the below image.
    Context: {context_text}
    Question: {user_question}
    """

    prompt_content = [{"type": "text", "text": prompt_template}]

    if len(docs_by_type["images"]) > 0:
        for image in docs_by_type["images"]:
            prompt_content.append(
                {
                    "type": "image_url",
                    "image_url": {"url": f"data:image/jpeg;base64,{image}"},
                }
            )

    return ChatPromptTemplate.from_messages(
        [
            HumanMessage(content=prompt_content),
        ]
    )


chain = (
    {
        "context": retriever | RunnableLambda(parse_docs),
        "question": RunnablePassthrough(),
    }
    | RunnableLambda(build_prompt)
    | ChatOpenAI(model="gpt-4o-mini")
    | StrOutputParser()
)

chain_with_sources = {
    "context": retriever | RunnableLambda(parse_docs),
    "question": RunnablePassthrough(),
} | RunnablePassthrough().assign(
    response=(
        RunnableLambda(build_prompt)
        | ChatOpenAI(model="gpt-4o-mini")
        | StrOutputParser()
    )
)
```

```python
response = chain_with_sources.invoke(
    "estan cosiderados los talleres inofensivos en el cuadro 2A"
)

print("Response:", response['response'])

print("\n\nContext:")
for text in response['context']['texts']:
    print("Page number: ", text.metadata.page_number)
    print(text.text)
    print("\n" + "-"*50 + "\n")
```
